```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
sns.set_style('ticks')
sns.set_style('whitegrid')
```

```python
df = pd.read_csv("custdata.tsv", sep="\t")
```

```python
combined_df = pd.concat([df["age"], df["income"], df["num.vehicles"]], axis=1)

corr_matrix = combined_df.corr(numeric_only=True)

print(corr_matrix)
```

```python
plt.figure(figsize=(10, 8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

data analysis - Correlation Matrix (custdata.tsv) age, income, and number of vehicles - Before Data Cleaning
----------------------------------
