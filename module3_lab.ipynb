Cleaning Log VII

1. Fill the missing prices using the 'median' statistic that resists skewed prices. 
2. Print the fill value I used and record the fill value for your cleaning log.
3. I will show that there are 0 missing prices afterwards.

```python
fill_value = df["price"].median()
print(f'The fill value represented by the medial of the price column is {fill_value}.')
df["price"] = df["price"].fillna(fill_value)
df.isna().sum()
```

Cleaning Log VIIi

Here, I show that there are 8 duplicates in the data frame before the .drop_duplicates() function. 

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
```

Cleaning Log X

Here, I pad the missing digits in the entries per the 'zip' column. I used the parameter 5 since the zip code has 5 digits in our data. I then printed the data frame to show the results of padding, when comparing with the other data frames printed earllier.

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
print(df)
```

Cleaning Log XI

Since there were mixed formatting in the 'date' column of our data, I used the .to_dateframe() function and its parameters to make the formatting uniform.

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
print(df)
```
