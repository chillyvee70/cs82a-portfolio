Cleaning Log I

Import libraries (panda & seaborn)

load .csv file containing data and store in variable, 'df'

```python
import pandas as pd
import seaborn as sns

df = pd.read_csv("sales_clean.csv")
```

Cleaning Log II

Compute mean, median, and standard deviation for price and for qty.

```python
print('========================  price  =============================')
print(f'The mean of the prices in our data is {df["price"].mean()}')
print(f'The median of the prices in our data is {df["price"].median()}')
print(f'The standard deviation of the prices in our data is {df["price"].std()}')
print('\n')
print('========================  quantity  =============================')
print(f'The mean of the quatity in our data is {df["qty"].mean()}')
print(f'The median of the quatity in our data is {df["qty"].median()}')
print(f'The standard deviation of the quantity in our data is {df["qty"].std()}')
```

Data Analysis for Mean, Median, and Std. Deviation of Price:
=============================================================
Since the price mean sits noticeably above the price median; that gap is seeded skew. Since there is a skew in the data distribution, there are more outliers that leads to a misreresenation of the data when considering the mean vs. median for describing a set of data. For price I would select the median to minimize the affect of outlying data points on our analysis.
