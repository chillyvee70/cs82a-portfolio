Cleaning Log I

```python
import pandas as pd
import seaborn as sns

df = pd.read_csv("allbacks.csv")
df
```

Cleaning Log II

```python
coverSort = df.sort_values("cover", axis=0, ascending=True, na_position='last')
coverSort
```

Tne median of 'cover' is between 7 and 8.

Cleaning Log III

```python
print(f'The mean weight of all books is {df['weight'].mean()} grams.')
```

Cleaning Log IV

```python
hbBooks = coverSort.head(7)
hbBooks
```

Cleaning Log V

```python
print(f'The mean weight of all hardback books is {hbBooks['weight'].mean()} grams.')
```

Cleaning Log VII

```python
pbBooks = coverSort.tail(8)
pbBooks
```

Cleaning Log VIII

```python
print(f'The mean weight of all paperback books is {pbBooks['weight'].mean()} grams.')
```

Cleaning Log IX

```python
print(f'The median weight of all books is {df['weight'].median()} grams.')
print(f'The median weight of all hardback books is {hbBooks['weight'].median()} grams.')
print(f'The median weight of all paperback is {pbBooks['weight'].median()} grams.')
```
