Cleaning Log I

Import libraries (panda & seaborn)

load .csv file containing data and store in variable, 'df'

```python
import pandas as pd
import seaborn as sns

df = pd.read_csv("truck.csv")
df
```

Cleaning Log II

Compute mean, median, and standard deviation for price and for qty.

```python
print('========================  price  =============================')
print(f'The mean price in our data is {df["X"].mean()}')
print(f'The median price in our data is {df["X"].median()}')
print(f'The price standard deviation in our data is {df["X"].std()}')
print('\n')
print('========================  best price  =============================')
print(f'The mean best price in our data is {df["Y"].mean()}')
print(f'The median best price in our data is {df["Y"].median()}')
print(f'The best price standard deviation in our data is {df["Y"].std()}')
```

Data Analysis for Mean, Median, and Std. Deviation of price & best price:
==========================================================================
Since the price mean sits noticeably above the price median; that is normal distribution for both sets of data. The shapes of the distribution are similar as indicated by the standard deviation closeness in value for both data sets.
