One markdown note on a MySQL-to-SQLite difference:
=============
The method of inserting rows is different. For the MySQL operation you have to use the keyword VALUES before specifying the row you intend to insert into the table, whereas in the SQLite operation no such keyword is needed. The one keyword in the statement is execute that is analogous to INSERT in the former's execution statement to perform the population of the tables.
