```python
temp = [10, 16, 23, 28, 14, 22, 30]
```

```python
for check in temp:                                                     
    if (check < 15):
        print(f'{check} is categorized as a Cold temperature.')
    elif (15 < check and check < 25):
        print(f'{check} is categorized as a Warm temperature.')
    else:
        print(f'{check} is categorized as a Hot temperature.')

```
