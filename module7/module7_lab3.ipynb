Import Necesssary Libraries
==

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
from sklearn.tree import plot_tree
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
```

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

Set Notebook Environmental Variables
==

```python
sns.set_style('ticks')
sns.set_style('whitegrid')
```

Load titanic_train.csv & titanic_test.csv
====

```python
df = pd.read_csv("titanic_train.csv")
df1 = pd.read_csv("titanic_test.csv")
df2 = pd.read_csv("titanic_testb.csv")
```

```python
df.head(10)
```

```python
df.isna().sum()
```

```python
sns.heatmap(df.isnull())
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
fill_value = df["Age"].mean()
print(f'The fill value represented by the mean of the price column is {fill_value}.')
df["Age"] = df["Age"].fillna(fill_value)
df.isna().sum()
```

```python
sns.heatmap(df.isnull())
```

```python
df["Sex_encoded"] = df["Sex"].map({"male": 0, "female": 1})
```

```python
X = df[[
    "Age",
    "Fare",
    "Sex_encoded"
]]
y = df["Survived"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
len(X_train)
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train,y_train)
test_accuracy = knn.score(X_test, y_test)
print("Test Accuracy:", test_accuracy)
training_accuracy = knn.score(X_train, y_train)
print("Training Accuracy:", training_accuracy)
```

Overfitting Analysis
===

There isn't a large gap between train vs. test scores this means the model learned. 
==

```python
tree_lrnd1 = DecisionTreeClassifier(max_depth=3, random_state=42)
tree_lrnd1.fit(X_train, y_train)

print("Test Accuracy:", tree_lrnd1.score(X_test, y_test))

plt.figure(figsize=(14, 8))
plot_tree(
    tree_lrnd1,
    feature_names=X_train.columns,
    class_names=[str(c) for c in tree_lrnd1.classes_],
    filled=True,
    rounded=True
)
plt.show()
```
