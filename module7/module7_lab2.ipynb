Load ratings.tsv

```python
df = pd.read_csv("ratings.tsv", sep="\t")
```

```python
df.head(10)
```

Features, target, and the split
===

```python
X = df[[
    "food",
    "ambience",
    "service"
]]
y = df["rating"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
len(X_train)
```

scikit-learn model: fit, predict, score
==

```python
model = LinearRegression()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
print(model.coef_, model.intercept_)
```

What drives a restaurant’s overall rating?
==

With model.coef_ function we get one slope per feature. The coefficients, representing the slope for each feature-respectively-are 0.95676835 for "food", 0.07401426 for "ambience", and 0.20380826 for "service". food ranks first, amboience ranks second and service ranks third. Considering the highest slope, representing a unit change in X will represent a higher unit change in rating-representing a customer's satisfaction.
