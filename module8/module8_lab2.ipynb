```python
df = pd.read_csv("lifecyclesavings.csv")

display(df.head())
```

```python
X = df[[
    "pop15",
    "pop75",
    "dpi",
    "ddpi"
]]
y = df["sr"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
len(X_train)
```

```python
knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(X_train,y_train)
test_accuracy = knn.score(X_test, y_test)
print("Test Accuracy:", test_accuracy)
training_accuracy = knn.score(X_train, y_train)
print("Training Accuracy:", training_accuracy)
```

Overfitting is possible due to large gap in the above values.

```python
from sklearn.tree import DecisionTreeRegressor, plot_tree
import matplotlib.pyplot as plt

# Create a decision tree regression model
tree_lrnd1 = DecisionTreeRegressor(
    max_depth=3,
    random_state=42
)

# Train the model
tree_lrnd1.fit(X_train, y_train)

# Evaluate model performance
print("Test R² Score:", tree_lrnd1.score(X_test, y_test))
print("Training R² Score:", tree_lrnd1.score(X_train, y_train))

# Display decision tree
plt.figure(figsize=(14, 8))

plot_tree(
    tree_lrnd1,
    feature_names=X_train.columns,
    filled=True,
    rounded=True,
    fontsize=10
)

plt.show()
```
