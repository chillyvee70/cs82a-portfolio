```python
fill_value2 = df["flipper_length_mm"].mean()
#print(f'The fill value represented by the mean of the price column is {fill_value2}.')
df["flipper_length_mm"] = df["flipper_length_mm"].fillna(fill_value2)
df.isna().sum()
```
